#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

```python
# 주민등록번호 => 뒷자리를 모두 * 문자로 변경

# 파이썬에서 정규식 지원하는 모듈

import re

dir(re)

# re.compile() : 정규식을 작성
p = re.compile("[abc]")

# 패턴을 특정 문자열이 포함하고 있는가?
print(p.match("abc"))
print(p.match("dude"))
```

| 메서드 | 목적 |
|-----|----|
|match | 문자열의 처음부터 정규식과 매치되는지 조사|
|search | 문자열 전체를 검색하여 정규식과 매치되는지 조사|
|findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴|
|finditer | 정규식과 매치되는 모든 문자열(substirng)을 이터레이터 객체로 리턴|

```python
m = p.match("a")

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"패턴 시작,끝 : {m.span()}")

if m:
    print(f"Match found {m.group()}")
else:
    print("No Match")
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|-----|----|
|\d|숫자와 매치[0-9] 와 동일한 표현식|
|\D|숫자가 아닌 것과 매치 [^0-9] 와 동일한 표현식|
|\s|화이트스페이스(스페이스,탭 줄바꿈 등)문자와 매치 [빈칸\t\n\r\v]과 동일한 표현식|
|\S|화이트스페이스(스페이스,탭 줄바꿈 등)문자가 아닌 것과 매치 [^빈칸\t\n\r\v]과 동일한 표현식|
|\w|문자 + 숫자와 매치 [a-zA-Z0-9_] 와 동일한 표현식|
|\W|문자 + 숫자가 아닌 것과 매치 [^a-zA-Z0-9_] 와 동일한 표현식|


```python
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))

# 원본 문자열에서 0~9가 아닌 문자 찾기
#p = re.compile("[^0-9]+")
p = re.compile(r"\D+")
print(p.search("012 abc"))

# 공백, 탭, 엔터 문자 찾기
#p = re.compile("[ \t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.search("012 abc"))

# 공백, 탭, 엔터 아닌 문자 찾기
#p = re.compile("[^ \t\n\r\f\v]+")
p = re.compile(r"[\S+]")
print(p.search("012 abc"))


# 대소문자,숫자,_포함한 문자 찾기
#p = re.compile("[a-zA-Z0-9_]+")
p = re.compile(r"\w+")
print(p.search("012 abc"))

# 대소문자,숫자,_가 아닌 문자 찾기
#p = re.compile("[^a-zA-Z0-9_]+")
p = re.compile(r"\W+")
print(p.search("012 abc"))


```

```python
# re.I : ignorecase

# 영어가 포함되어 있는지 확인
p = re.compile("[a-z]+",re.I)
print(p.match("python"))
print(p.match("Python"))
print(p.match("PYTHON"))
```

```python
# re.M

data="""python one
life is too short
python two
you need python
python three
"""

# ^ : 시작하는
# python 으로 시작하고, 스페이스문자하나 영어,숫자,_ 인 문자 찾아줘 
p=re.compile(r"^python\s\w+",re.M)
print(p.findall(data))
```

```python
# | : or

p = re.compile("Crow|Servo")
m = p.match("CrowHello")
print(m)

m = p.match("ServoHello")
print(m)
```

```python
# ^ : 시작하는, $ : 끝나는

p = re.compile("^Life")
m = p.search("Life is too short")
print(m)

m = p.search("My Life")
print(m)

p = re.compile("short$")
m = p.search("Life is too short")
print(m)
```

```python
# () : 그루핑

# ABCABC 문자열이 반복되는지 조사

p = re.compile("(ABC)+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
print(m.group(1))
```

```python
with open("./data/log.txt","r",encoding="utf-8")as f:
    logs = f.read()
#print(logs)
#enter 기준으로 분리
logs= logs.splitlines()

error, info = 0, 0
pattern = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2})\s+(?P<level>INFO|ERROR)\s+(?P<message>.+)")

for log in logs:
    m = pattern.search(log)

    if not m:
        continue

    date = m.group('date')
    level = m.group('level')
    message = m.group('message')

    if level == "INFO":
        info += 1
    elif level == "ERROR":
        error += 1
        print(message)
        daily_error[date] = daily_error.get(date,0) + 1


print("ERROR 목록")
print("-"* 20)

    

print(f"INFO : {info}")
print(f"ERROR : {len(err_logs)}")

for key, value in daily_error.items():
    print(f"{key} : {value}")
```
