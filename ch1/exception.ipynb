```python
# 예외 클래스 생성 가능
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다."

def say_nick(nick):
    if nick =="바보":
        raise MyError()
    print(nick)




try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
```

```python
# 문제 1. 랜덤 비밀번호 생성기
from random import *
# 사용자로부터 생성할 비밀번호의 길이를 입력받아 랜덤 비밀번호를 생성하는 프로그램 작성
# choice()
chars = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789!@#$%"
# 비밀번호 최소길이 4 이상
# 숫자가 아닌 값을 입력하면 예외처리

while True:
    try:
        pass1 = int(input("비밀번호를 길이를 입력하세요 (최소4)"))
        if int(pass1) < 4:
            print("비밀번호 길이를 다시 입력해주세요 (최소4)")
            continue
#        else:
#            for c in range(pass1):
#                print(choice(chars),end="")
        break
    except ValueError:
            print("잘못된 입력입니다.")
#password = []
#for c in range(pass1):
#        password.append(choice(chars))
password = [ choice(chars) for _ in range(pass1) ]
# print(f"랜덤 비밀번호 : {password}")
# 리스트 => 문자열로 변경
print(f"랜덤 비밀번호 : {"".join(password)}")
```

```python
# 나이 계산 프로그램
# 생년월일을 입력받아 현재 나이를 계산하는 프로그램 작성
# 입력형식 2002-08-25 => 날짜 형식 strptime
import datetime


now = datetime.datetime.now()

try:
    #생년월일 입력
    birth_str = input("생년월일 입력(YYYY-MM-DD) ")

    birth = datetime.datetime.strptime(birth_str,"%Y-%m-%d")

    # 2026 - 2001
    # now 년도만 추출

    age = now.year - birth.year
    if (now.month, now.day) < (birth.month, birth.day):
        age -= 1

    print("나이",age)
except:
    print("날짜 형식이 올바르지 않습니다.")







#birth1 = input("생년월일을 입력하세요 예시) 2001-08-25")

#birth1 = datetime.datetime.strptime(str(birth1),"%Y-%m-%d")

#print(birth1)
```
