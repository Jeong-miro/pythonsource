```python
# 학생 5명의 점수를 입력받아 다음 정보를 출력하는 프로그램

# 출력내용
# 최고점수max, 최저점수min, 평균점수(소수 둘째 자리)sum/len, 점수를 오름차순으로 출력sorted, 등수와 함께 출력enumerate
numbers = []

for i in range(5):
    jumsu = int(input(f"{i+1} 번째 점수 입력 : "))
    numbers.append(jumsu)
print(numbers)


print("최고점수 :",max(numbers))
print("최저점수 :",min(numbers))
print("평균점수 :",sum(numbers)/len(numbers))
print("오름차순 정렬 :",sorted(numbers))

# 등수
sorted_desc = sorted(numbers,reverse=True)
for idx, jumsu in enumerate(sorted_desc,1):
    print(f"{idx} : {jumsu}")
#averstat = sum(list1)/int(len(list1))
#print(sorted(enumerate(list1)))
```

```python
# 문자열을 입력받아 다음 정보를 출력
# 출력내용 : 입력한 문자, 각 문자의 유니코드 값, 가장 큰 문자, 가장 작은 문자

# ord,chr,max,min,list
# 문자열 입력 : Python
# P -> 80
# y -> 121
#
# 가장 큰 문자 : y
# 가장 큰 문자 : P
a = input("문자 입력")
list2 = [a]
for i in range(len(a)):
    a2 = ord(a[i])
    print(f"{a[i]} -> {a2}")
print(f"가장 큰 숫자 : {ord(max(a))}")
print(f"가장 작은 숫자 : {ord(min(a))}")

word = input("영어 문자열 입력 :")
for ch in word:
    print(f"{ch} -> {ord(ch)}")

print(f"가장 큰 숫자 : {ord(max(word))}")
print(f"가장 작은 숫자 : {ord(min(word))}")
```

```python
# 사용자로부터 숫자 10개 입력받기

# 출력
# 입력리스트, 중복제거, 오름차순, 튜플변환, 집합 크기

# 3 5 2 5 7 ~~

# 원본 [3,5,2,5~~]
# 중복제거 {1,2,3,4} set()
# 정렬 sorted()
# 튜플 tuple()
# 집합 크기 : len() 

#number1 = input("숫자 10개 입력 예시) 1,2,3,4...")
#print(number1)
#print(set(number1))
#print(sorted(number1))
#print(tuple(number1))
#print(f"집합 크기 : {len(number1)}")

numbers = []
for i in range(10):
    num = int(input(f"숫자 입력 : "))
    numbers.append(num)

print(f"원본 리스트 : {numbers}")
unique_numbers = list(set(numbers))
print(f"중복제거 : {unique_numbers}")
sorted_numbers = sorted(unique_numbers)
print(f"오름차순 정렬 : {sorted_numbers}")
print(f"리스트 => 튜플 : {tuple(sorted_numbers)}")
print(f"중복 제거 후 집합 크기 {len(unique_numbers)}")
```
