### python 과 데이터베이스 연동
내장 DB 연동 : SQLite

```python
import sqlite3

# isolation_level : 트랜잭션
# None = autocommit 중지
conn = sqlite3.connect("./test.db",isolation_level=None)

# sql 을 db에 전달하고 결과를 받아오는 객체
cursor = conn.cursor();
```

```python
# 테이블 생성
sql = """create table if not exists users(id integer primary key,
username text,
email text,
phone text,
website text,
regdate text
)
"""

cursor.execute(sql)
```

```python
from datetime import datetime
now = datetime.now()
nowDateTime = now.strftime("%Y-%m-%d %H:%M:%S")
nowDateTime
```

```python
# insert 구문 : 튜플 형태로 
sql = "insert into users(id,username,email,phone,website,regdate)" \
"values(1,'kim','kim@naver.com','010-1234-5678','kim.com',?)"

cursor.execute(sql,(nowDateTime,))
```

```python
sql = "insert into users(id,username,email,phone,website,regdate)" \
"values(?,?,?,?,?,?)"

cursor.execute(sql,(2,'park','park@naver.com','010-2345-5678','park.com',nowDateTime))
```

```python
user_list = [
    (3,'hong','hong@naver.com','010-2345-5678','hong.com',nowDateTime),
    (4,'cho','cho@naver.com','010-2345-5678','cho.com',nowDateTime),
    (5,'yoo','yoo@naver.com','010-2345-5678','yoo.com',nowDateTime),
]

cursor.executemany(sql,user_list)
```

```python
# 데이터 조회

sql = "select * from users"

cursor.execute(sql)
```

```python
cursor.fetchone()
```

```python
cursor.fetchmany(size=3)
```

```python
cursor.fetchall()
```

```python
sql = "select * from users order by id desc"

cursor.execute(sql)

for row in cursor.fetchall():
    print(row)
```

```python
sql = "select * from users where id =3 "

cursor.execute(sql)
cursor.fetchall()
```

```python
sql = "select * from users where id =? "

cursor.execute(sql,(4,))
cursor.fetchall()
```

```python
sql = "select * from users where id in (?,?) "
cursor.execute(sql,(3,5))
cursor.fetchall()
```

```python
sql = "update users set username=? where id=?"

cursor.execute(sql,('young',2))
```

```python
cursor.execute("select * from users where id=2")
cursor.fetchone()
```
