```python
from faker import Faker
fake = Faker(locale="ko_KR")
Faker.seed(4321)

print(fake.name())
print(fake.email())
print(fake.address())
print(fake.ssn())
print(fake.phone_number())
```

```python
# maru_member 50명 임의의 데이터를 이용해 삽입
# name birth phone joined_at
from faker import Faker
marulist = []
for i in range(50):
    datalist = [
    [fake.name(),
    fake.date_of_birth(minimum_age=20,maximum_age=30).strftime("%Y-%m-%d"),
    fake.phone_number(),
    fake.date_time_between(start_date="-4y",end_date="now").strftime("%Y-%m-%d"),
        ]
    ]
    marulist += datalist
print(marulist)

# 50명의 데이터를 리스트에 추가 
# 그 리스트를 executemany 를 통해 insert 구문으로 추가

sql = "insert into maru_member (name, birth, phone, joined_at) values (:1, :2, :3, :4)"
with oracledb.connect(user="python_user",password=password,dsn="localhost/xe") as conn:
    with conn.cursor() as cursor:
        cursor.executemany(sql,marulist)
        conn.commit()
```

```python
# maru_member 50명 임의의 데이터를 이용해 삽입
# name birth phone joined_at
from faker import Faker
data = []
for i in range(50):

    name = fake.name()
    birth = fake.date_of_birth(minimum_age=20,maximum_age=70).strftime("%Y-%m-%d")
    phone = fake.phone_number()
    joined_at = fake.date_time_between(start_date="-4y",end_date="now").strftime("%Y-%m-%d")
    data.append([name,birth,phone,joined_at])
print(data)

# 50명의 데이터를 리스트에 추가 
# 그 리스트를 executemany 를 통해 insert 구문으로 추가

sql = "insert into maru_member (name, birth, phone, joined_at) values (:1, :2, :3, :4)"
with oracledb.connect(user="python_user",password=password,dsn="localhost/xe") as conn:
    with conn.cursor() as cursor:
        cursor.executemany(sql,data)
        conn.commit()
```
